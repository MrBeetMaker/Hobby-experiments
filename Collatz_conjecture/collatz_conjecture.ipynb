### Collcatz Conjecture
* You are holding a number n
* If n is even, divide it by 2
* If n is odd, multiply by 3 and add 1

If we do this over and over, will we always end up at 1? That's Collatz Conjecture.

In this notebook, we will investigate it.

### Libraries

```python
import matplotlib.pyplot as plt
import numpy as np
```

### Functions

We can visualize the progressions for many different numbers using this function, which will be helpful later.

```python

def visualize_progressions(data: list[int], extra_info= None):

    max_cols = 5

    if len(data) >= max_cols:
        cols = max_cols
    else:
        cols = len(data) % max_cols

    rows = len(data) // cols

    if rows * cols < len(data):
        rows += 1

    width = 3.8*cols
    height = 2.7*rows

    if cols <= 2:
        width *= 1.7
        height *= 1.7
    elif cols == 3:
        width *= 1.5
        height *= 1.5

    fig = plt.figure(figsize= (width, height))
    axes = fig.subplots(nrows= rows, ncols= cols)

    if rows == 1 and cols == 1:                 # Make sure axes is always 2D for indicing
        axes = np.array([[axes]])
    elif rows == 1 or cols == 1:
        axes = np.array(axes).reshape(rows, cols)
    else:
        axes = np.array(axes)

    for i in range(rows * cols):

        row = i // cols
        col = i % cols

        if i >= len(data):
            fig.delaxes(axes[row][col])
            continue

        axes[row][col].plot(data[i])


        title = ""
        if extra_info:
            try:
                title += f"{extra_info[i]}"
            except Exception:
                pass

        axes[row][col].set_title(title)
        # axes[row][col].axis('off')

    fig.patch.set_facecolor("lightgray")

    fig.tight_layout()
```

This is the implementation for getting the next number in the sequence.

```python

def next(n: int)-> int:
    if n % 2 == 0:
        return n // 2
    else:
        return 3 * n + 1
```

We can test different numbers using this function. It stores the progression so we can visualize it later.

```python

def run_problem(n, max_iter = 10000):

    progression = list()

    iter = 0
    while n != 1 and iter < max_iter:
        n = next(n)

        progression.append(n)
        iter += 1

    return progression
```

Now, we can finally run the experiment

```python

titles = list()
progressions = list()

for i in range(3, 200):

    titles.append(i)
    progressions.append(run_problem(i))

    if progressions[-1][-1] != 1:
        print(f"{progressions[-1][0]} did not converge within {len(progressions[-1])} iterations.")
```

Here's a visualization of all of the progressions

```python

visualize_progressions(progressions, titles)
```
