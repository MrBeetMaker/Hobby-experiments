## Anomaly detection


Anomaly detection in helicopter:
- https://www.kaggle.com/datasets/nagasai524/anomaly-detection-in-helicopter


```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

source = "data"
output_folder = "data"
```

### Load the datasets
The dataset consists of three files with train data, test data and labels for the test data. All samples in the train data are without anomalies.

```python
# Load the dataset

train_path = f"{source}\\train_df.csv"
test_path =  f"{source}\\validation_df.csv"
labels_path = f"{source}\\dfvalid_groundtruth.csv"

df, test_df, labels = pd.read_csv(train_path), pd.read_csv(test_path), pd.read_csv(labels_path)
```

```python
# Rename first column

df.columns = ["seqID"] + list(df.columns[1:].astype(int))
test_df.columns = ["seqID"] + list(test_df.columns[1:].astype(int))

labels = labels.astype(int)

df
```

About the data:
- The training data contains 1677 sequences of normal flight data.
- Each row holds amplitude sampled at 1024Hz for 1 minute of a flight.
- There are 594 additional labled sequences for testing.

```python
print(f"There are {len(labels)} labels and {len(test_df)} test samples: " + "one label for each test sample." if len(labels) == len(test_df) else "some labels are missing.")

test_df
```

```python

fig, ax = plt.subplots(figsize=(7, 5))

nr_of_anomalies = sum(labels["anomaly"])
nr_of_normal = len(labels) - nr_of_anomalies

ax.set_title("Distribution of test data")
ax.bar((0.25, 0.75), (nr_of_normal, nr_of_anomalies), color='firebrick', width=0.25, zorder=10)
ax.set_xticks((0.25, 0.75))
ax.set_xticklabels(("Normal", "Anomaly"))
ax.set_xlim(-0, 1)
ax.set_ylabel("Frequency")
ax.grid(axis="y", zorder= 1)
fig.patch.set_facecolor('lightgrey')
```

```python

fig, ax = plt.subplots(figsize=(7, 6))

flattened_df = df.iloc[:, 1:].values.flatten()
# flattened_df.sort()

ax.set_title("Value distribution")
ax.grid(axis="y", zorder=1)
ax.plot(flattened_df, color='firebrick')

fig.patch.set_facecolor('lightgrey')

percentiles = np.percentile(flattened_df, [25, 75])
print(f"Value distribution:\n * Mean: {np.mean(flattened_df):.2f}\n * Std: {np.std(flattened_df):.2f}\n * Min: {np.min(flattened_df):.2f}\n * Max: {np.max(flattened_df):.2f}\n * 25th percentile: {percentiles[0]:.2f}\n * 75th percentile: {percentiles[1]:.2f}")
```

```python
train_arr = df.iloc[:, 1:].values.astype(np.float32)
test_arr = test_df.iloc[:, 1:].values.astype(np.float32)
```

```python
fig, ax = plt.subplots(figsize=(7, 6))

flattened_arr = train_arr.flatten()

ax.set_title("Value distribution of normalized data")
ax.grid(axis="y", zorder=1)
ax.plot(flattened_arr, color='firebrick')

fig.patch.set_facecolor('lightgrey')

percentiles = np.percentile(flattened_arr, [25, 75])
print(f"Value distribution of normalized data:\n * Mean: {np.mean(flattened_arr):.2f}\n * Std: {np.std(flattened_arr):.2f}\n * Min: {np.min(flattened_df):.2f}\n * Max: {np.max(flattened_df):.2f}\n * 25th percentile: {percentiles[0]:.2f}\n * 75th percentile: {percentiles[1]:.2f}")
```

### Save dataset as .npy
This allows us to read the dataset much more efficiently in the next part.

```python
train_location = f"{output_folder}\\helicopter_train.npy"
test_location = f"{output_folder}\\helicopter_test.npy"

np.save(train_location, train_arr)        # 0.4 GB instead of 2 GB
np.save(test_location, test_arr)

print(f"Saved dataset as '{train_location}' and '{test_location}'")
```
